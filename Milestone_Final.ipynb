**4.2 Answer:**  

The raw Zillow dataset required extensive preprocessing due to missing values, high-cardinality categorical variables, identifier fields, and a small number of constant or unusable features.

**Issues Identified**

- 20 features had more than 90% missing values (e.g., architectural style, basement area, pool-related features) and were removed. 

- 5 additional features with 77-89% missing values (fireplace count, three-quarter bathroom count, pool type, number of stories) were also dropped. 

- Identifier variables (parcel ID, census tract) and a constant feature (assessment year) were eliminated. 

- 35 samples with missing target values or more than 60% missing features (6 samples) were removed, resulting in a **final dataset of 77,572 samples.**

**Outliers**

- We identified 5,184 extreme tax values, 4,006 extreme square footage values, and 11,408 extreme lot sizes. These outliers were retained rather than removed. High-value properties represent a meaningful segment of the market, and removing them would limit the model's ability to generalize across the full price range.

**Imputation Strategy**

Missing values were handled using several approaches:

- **Constant imputation:** Where missing indicated absence (garage count, pool count, unit count)
- **Median imputation:** For 24 numeric features (square footage, bedrooms, bathrooms, latitude, longitude), using median to avoid distortion from skewed distributions
- **Mode imputation:** For categorical ID columns (regionidcity, regionidzip, fips) and object columns (property zoning, land use code)

**Processing alternatives Considered**

- Mean imputation was rejected due to the data's right-skewed distributions. 

- One-hot encoding was evaluated but would have expanded the feature space to over 2,900 columns, causing memory issues. 

- Frequency encoding was selected to preserve category information while keeping the feature set compact (only 28 columns, no added features). 

- Other complex imputation techniques thought of (KNN, iterative) but were rejected due to computational cost.

**Feature Engineering**

We created three new features:
- **has_pool** (binary from pool count)
- **has_garage** (binary from garage count)
- **age** (age = 2016 - year built)

**Data Leakage Prevention**

- All preprocessing steps, like imputation and encoding, were **fit** exclusively on the **training data** and **applied** to the **test set**.

**Final preprocessing pipeline Summary**

- Removed 26 features (identifiers, constant, >77% missing)

- Removed 41 samples

- Retained outliers

- Applied constant, median, and mode imputation

- Created **has_pool, has_garage, age** features

- Frequency encoded categorical variables

- Fit preprocessing using only training data

- **Final dataset: 28 features, 77,572 samples, no missing values**

These preprocessing decisions produced a cleaner, lower-dimensional dataset that retained the most informative property characteristics while reducing noise, minimizing sparsity, and improving the suitability of the data for subsequent machine learning models.

**4.4 Answer:**

We selected three diverse models to establish baselines and evaluate performance:

1. **Linear Regression:** A simple, interpretable baseline that assumes linear relationships between features and the target. We included **Ridge Regression** (an L2-regularized version) during hyperparameter tuning to test whether regularization could improve performance.

2. **Random Forest:** An ensemble of decision trees that captures non-linear relationships and interactions naturally. It is robust to irrelevant features and provides feature importance estimates.

3. **Decision Tree:** A single tree-based model that is interpretable but prone to overfitting. It served as a baseline for comparing ensemble methods.

**Model Evaluation:** All models were evaluated using repeated 5-fold cross-validation (5 repeats) on the training set. This provided stable estimates of generalization performance. We compared models based on mean CV MAE and standard deviation, prioritizing lower MAE and lower variance.

**Hyperparameter Tuning:** We used validation curves to identify promising hyperparameter ranges, followed by focused searches:

- Ridge Regression: GridSearchCV over alpha values

- Random Forest: RandomizedSearchCV over n_estimators, max_depth, min_samples_split, min_samples_leaf, and max_features

- Decision Tree: GridSearchCV over max_depth, min_samples_split, min_samples_leaf, and max_features

**Trade-offs:** Random Forest offered the best balance of predictive performance and stability. While it is less interpretable than Linear Regression, its superior accuracy justified the trade-off. It also required minimal feature engineering (no scaling, robust to irrelevant features), reducing preprocessing complexity.

**Test Set Reservation:** The test set was held out throughout all modeling decisions and used only once for final evaluation. This ensured an unbiased estimate of generalization performance.

**Final Model Selection:** Random Forest with tuned hyperparameters achieved the lowest CV MAE (\$190,741.31) and high stability (std: \$3,127.89). It was selected as the final model.
